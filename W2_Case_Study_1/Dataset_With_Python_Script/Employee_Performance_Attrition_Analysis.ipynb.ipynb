Check Employee ID Coverage Across Datasets
We need to determine which employee IDs in Attrition Data actually exist in Employee Data.

```python
employee_ids = set(employee_df["Employee_ID"])
attrition_ids = set(attrition_df["employee_ID"])
performance_ids = set(performance_df["Employee_ID"])

print("Attrition IDs found in Employee Data:",
      len(attrition_ids.intersection(employee_ids)))

print("Attrition IDs NOT found in Employee Data:",
      len(attrition_ids - employee_ids))

print("Performance IDs found in Employee Data:",
      len(performance_ids.intersection(employee_ids)))

print("Performance IDs NOT found in Employee Data:",
      len(performance_ids - employee_ids))
```

Check Missing Employee IDs

```python
missing_attrition_ids = attrition_ids - employee_ids
missing_performance_ids = performance_ids - employee_ids

print("Missing Attrition Employee IDs:")
print(missing_attrition_ids)

print("\nMissing Performance Employee IDs:")
print(missing_performance_ids)
```

employee IDs in the child datasets exist in the Employee dataset.

Investigate Repeated Attrition Records

```python
repeated_ids = repeated_attrition_ids.index

repeated_attrition_records = attrition_df[
    attrition_df["employee_ID"].isin(repeated_ids)
].sort_values("employee_ID")

display(repeated_attrition_records)
```

```python
# Check conflicts 
attrition_conflicts = (
    repeated_attrition_records
    .groupby("employee_ID")
    .agg(
        record_count=("employee_ID", "size"),
        attrition_values=("attrition", lambda x: list(x)),
        exit_scores=("Exit_Interview_Score", lambda x: list(x))
    )
    .reset_index()
)

display(attrition_conflicts)
```

```python
# Performing cleaning in copy Dataset
employee_clean = employee_df.copy()
attrition_clean = attrition_df.copy()
performance_clean = performance_df.copy()

print("Working copies created successfully.")
```

```python
# Standardize Employee Column Names

employee_clean = employee_clean.rename(columns={
    "first_name": "First_Name",
    "last_name": "Last_Name",
    "gender": "Gender_Detailed",
    "Gender": "Gender_Group"
})

print(employee_clean.columns.tolist())
```

```python
# Standardize Attrition Column Names
attrition_clean = attrition_clean.rename(columns={
    "employee_ID": "Employee_ID"
})

print(attrition_clean.columns.tolist())
```

```python
attrition_clean = attrition_clean.rename(columns={
    "attrition": "Attrition"
})

print(attrition_clean.columns.tolist())
```

```python
performance_clean.columns.tolist()
```

## Remove Only Exact Duplicate Attrition Rows

```python
before_duplicates = len(attrition_clean)

attrition_clean = attrition_clean.drop_duplicates()

after_duplicates = len(attrition_clean)

print("Records before duplicate removal:", before_duplicates)
print("Records after duplicate removal:", after_duplicates)
print("Exact duplicate records removed:", before_duplicates - after_duplicates)
```

```python
print("Employee Data")
print("Rows:", len(employee_clean))
print("Duplicate rows:", employee_clean.duplicated().sum())

print("\nAttrition Data")
print("Rows:", len(attrition_clean))
print("Duplicate rows:", attrition_clean.duplicated().sum())

print("\nPerformance Data")
print("Rows:", len(performance_clean))
print("Duplicate rows:", performance_clean.duplicated().sum())
```

```python
print("Employee Data Missing Values:")
print(employee_clean.isnull().sum().sum())

print("\nAttrition Data Missing Values:")
print(attrition_clean.isnull().sum().sum())

print("\nPerformance Data Missing Values:")
print(performance_clean.isnull().sum().sum())
```

```python
# Validate the Employee Relationships Again

employee_ids = set(employee_clean["Employee_ID"])
attrition_ids = set(attrition_clean["Employee_ID"])
performance_ids = set(performance_clean["Employee_ID"])

print("Attrition IDs not found in Employee Data:",
      len(attrition_ids - employee_ids))

print("Performance IDs not found in Employee Data:",
      len(performance_ids - employee_ids))
```

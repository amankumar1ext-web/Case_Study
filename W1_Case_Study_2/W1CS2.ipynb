Missing Sales Amount

```python
sales["SalesAmount"] = sales["SalesAmount"].fillna(
sales["SalesAmount"].median()
)
```

```python
print(sales.isnull().sum())
print("______________________")
```

Missing Quantity

```python
sales["Quantity"] = sales["Quantity"].fillna(
    sales["Quantity"].median()
)
```

##Gender Standardization


```python
customers["Gender"] = customers["Gender"].replace({
    "M":"Male",
    "Male ":"Male",
    "F":"Female",
    "Female ":"Female"
})
```

```python
customers.head(5)
```

```python
customers.nunique()
```

Region Standardization

```python
customers["Region"] = customers["Region"].replace({
    "Ohho":"Ohio",
    "OHIO":"Ohio",
    "NY":"New York",
    "New Yorkk":"New York",
    "Nw York":"New York"
})
```

```python
customers.head(5)
```

```python
customers["Region"] = customers["Region"].fillna("Unknown")
```

```python
sales = sales.drop_duplicates()
customers = customers.drop_duplicates()
products = products.drop_duplicates()
```

```python
sales.count()
```

```python
print(sales["SalesAmount"].min())
print(sales["SalesAmount"].max())
sales["SalesAmount"].head()
```

```python
Q1 = sales["SalesAmount"].quantile(0.25)
Q3 = sales["SalesAmount"].quantile(0.75)
```

```python
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = sales[
    (sales["SalesAmount"] < lower) |
    (sales["SalesAmount"] > upper)
]

print(outliers.shape)
```

```python
print(outliers)
```

```python
sales["Timestamp"] = pd.to_datetime(
    sales["Timestamp"]
)

sales["Year"] = sales["Timestamp"].dt.year

sales["Month"] = sales["Timestamp"].dt.month

sales["MonthName"] = sales["Timestamp"].dt.month_name()
```

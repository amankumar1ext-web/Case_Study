```python
# INVENTORY REORDER ANALYSIS

inventory_data["Reorder_Status"] = (
    inventory_data["Stock_Level"]
    <= inventory_data["Reorder_Level"]
)

print("REORDER STATUS")
print(
    inventory_data["Reorder_Status"]
    .value_counts()
)
```

```python
print("\nREORDER STATUS (%):")
print(
    inventory_data["Reorder_Status"]
    .value_counts(normalize=True).mul(100).round(2)
)
```

```python
print("\nPRODUCT-WISE REORDER ANALYSIS")

reorder_by_product = (
    inventory_data
    .groupby("Product_ID")
    .agg(
        Total_Records=("Product_ID", "count"),
        Reorder_Records=("Reorder_Status", "sum"),
        Average_Stock=("Stock_Level", "mean"),
        Average_Reorder_Level=("Reorder_Level", "mean")
    )
    .reset_index()
)

reorder_by_product["Reorder_Percentage"] = (
    reorder_by_product["Reorder_Records"]
    / reorder_by_product["Total_Records"] * 100
).round(2)

display(
    reorder_by_product.sort_values(
        "Reorder_Percentage",
        ascending=False
    )
)
```

## EDA


```python
# SALES ANALYSIS BY PRODUCT

sales_by_product = (
    sales_data
    .groupby("Product_ID")
    .agg(
        Total_Quantity_Sold=("Quantity_Sold", "sum"),
        Total_Revenue=("Revenue", "sum"),
        Average_Quantity=("Quantity_Sold", "mean"),
        Average_Revenue=("Revenue", "mean"),
        Number_of_Sales=("Sale_ID", "count")
    )
    .reset_index()
)

display(
    sales_by_product.sort_values(
        "Total_Revenue",
        ascending=False
    )
)
```

```python
# SALES ANALYSIS BY STORE

sales_by_store = (
    sales_data
    .groupby("Store_ID")
    .agg(
        Total_Quantity_Sold=("Quantity_Sold", "sum"),
        Total_Revenue=("Revenue", "sum"),
        Number_of_Sales=("Sale_ID", "count")
    )
    .reset_index()
)

display(
    sales_by_store.sort_values(
        "Total_Revenue",
        ascending=False
    )
)
```

```python
# PURCHASE ORDER ANALYSIS

po_by_product = (
    purchase_orders_data
    .groupby("Product_ID")
    .agg(
        Total_Orders=("Order_ID", "count"),
        Total_Quantity_Ordered=("Quantity", "sum"),
        Average_Order_Quantity=("Quantity", "mean")
    )
    .reset_index()
)

display(
    po_by_product.sort_values(
        "Total_Quantity_Ordered",
        ascending=False
    )
)
```

```python
# SUPPLIER ANALYSIS

supplier_analysis = (
    suppliers_data
    .groupby("Order_Frequency")
    .agg(
        Supplier_Count=("Supplier_ID", "count"),
        Average_Lead_Time=("Lead_Time (days)", "mean"),
        Minimum_Lead_Time=("Lead_Time (days)", "min"),
        Maximum_Lead_Time=("Lead_Time (days)", "max")
    )
    .reset_index()
)

display(supplier_analysis)
```

```python
# INVENTORY VS SALES ANALYSIS

inventory_summary = (
    inventory_data
    .groupby("Product_ID")
    .agg(
        Average_Stock=("Stock_Level", "mean"),
        Average_Reorder_Level=("Reorder_Level", "mean"),
        Reorder_Records=("Reorder_Status", "sum"),
        Inventory_Records=("Product_ID", "count")
    )
    .reset_index()
)

sales_summary = (
    sales_data
    .groupby("Product_ID")
    .agg(
        Total_Quantity_Sold=("Quantity_Sold", "sum"),
        Total_Revenue=("Revenue", "sum"),
        Average_Quantity_Sold=("Quantity_Sold", "mean")
    )
    .reset_index()
)

inventory_sales = inventory_summary.merge(
    sales_summary,
    on="Product_ID",
    how="inner"
)

inventory_sales["Reorder_Percentage"] = (
    inventory_sales["Reorder_Records"]
    / inventory_sales["Inventory_Records"] * 100
).round(2)

display(
    inventory_sales.sort_values(
        "Reorder_Percentage",
        ascending=False
    )
)
```

```python
# INVENTORY BY STORE

inventory_by_store = (
    inventory_data
    .groupby("Store_ID")
    .agg(
        Average_Stock=("Stock_Level", "mean"),
        Average_Reorder_Level=("Reorder_Level", "mean"),
        Reorder_Records=("Reorder_Status", "sum"),
        Total_Records=("Product_ID", "count")
    )
    .reset_index()
)

inventory_by_store["Reorder_Percentage"] = (
    inventory_by_store["Reorder_Records"]
    / inventory_by_store["Total_Records"] * 100
).round(2)

display(
    inventory_by_store.sort_values(
        "Reorder_Percentage",
        ascending=False
    )
)
```

```python
# INVENTORY BY WAREHOUSE

inventory_by_warehouse = (
    inventory_data
    .groupby("Warehouse_ID")
    .agg(
        Average_Stock=("Stock_Level", "mean"),
        Average_Reorder_Level=("Reorder_Level", "mean"),
        Reorder_Records=("Reorder_Status", "sum"),
        Total_Records=("Product_ID", "count")
    )
    .reset_index()
)

inventory_by_warehouse["Reorder_Percentage"] = (
    inventory_by_warehouse["Reorder_Records"]
    / inventory_by_warehouse["Total_Records"] * 100
).round(2)

display(
    inventory_by_warehouse.sort_values(
        "Reorder_Percentage",
        ascending=False
    )
)
```

# Data Model Tables

```python
# CREATE DIMENSION TABLES

# Product Dimension
dim_product = pd.DataFrame({
    "Product_ID": sorted(
        sales_data["Product_ID"].unique()
    )
})

# Store Dimension
dim_store = pd.DataFrame({
    "Store_ID": sorted(
        sales_data["Store_ID"].unique()
    )
})

# Warehouse Dimension
dim_warehouse = pd.DataFrame({
    "Warehouse_ID": sorted(
        inventory_data["Warehouse_ID"].unique()
    )
})

# Supplier Dimension
dim_supplier = suppliers_data.copy()

dim_supplier = dim_supplier.rename(
    columns={
        "Lead_Time (days)": "Lead_Time_Days"
    }
)

# Remove duplicate supplier records if any
dim_supplier = dim_supplier.drop_duplicates(
    subset=["Supplier_ID"]
)

print("Dim_Product:", dim_product.shape)
print("Dim_Store:", dim_store.shape)
print("Dim_Warehouse:", dim_warehouse.shape)
print("Dim_Supplier:", dim_supplier.shape)
```

```python
# CREATE DATE DIMENSION

all_dates = pd.concat([
    inventory_data["Last_Updated"],
    sales_data["Sale_Date"],
    purchase_orders_data["Order_Date"],
    purchase_orders_data["Arrival_Date"]
])

min_date = all_dates.min()
max_date = all_dates.max()

date_range = pd.date_range(
    start=min_date,
    end=max_date,
    freq="D"
)

dim_date = pd.DataFrame({
    "Date": date_range
})

dim_date["Year"] = dim_date["Date"].dt.year
dim_date["Quarter"] = (
    "Q" + dim_date["Date"].dt.quarter.astype(str)
)
dim_date["Month_Number"] = dim_date["Date"].dt.month
dim_date["Month_Name"] = dim_date["Date"].dt.month_name()
dim_date["Day"] = dim_date["Date"].dt.day
dim_date["Day_of_Week"] = dim_date["Date"].dt.dayofweek + 1
dim_date["Day_Name"] = dim_date["Date"].dt.day_name()

print("Date range:")
print(dim_date["Date"].min())
print(dim_date["Date"].max())

print("\nDate dimension shape:")
print(dim_date.shape)

display(dim_date.head())
```

```python
# FACT SALES

fact_sales = sales_data[
    [
        "Sale_ID",
        "Product_ID",
        "Store_ID",
        "Sale_Date",
        "Quantity_Sold",
        "Revenue"
    ]
].copy()

print("Fact Sales:", fact_sales.shape)
display(fact_sales.head())
```

```python
# FACT INVENTORY

fact_inventory = inventory_data[
    [
        "Product_ID",
        "Store_ID",
        "Warehouse_ID",
        "Last_Updated",
        "Stock_Level",
        "Reorder_Level"
    ]
].copy()

print("Fact Inventory:", fact_inventory.shape)
display(fact_inventory.head())
```

```python
# FACT PURCHASE ORDERS

fact_purchase_orders = purchase_orders_data[
    [
        "Order_ID",
        "Product_ID",
        "Supplier_ID",
        "Order_Date",
        "Quantity",
        "Arrival_Date"
    ]
].copy()

print("Fact Purchase Orders:",
      fact_purchase_orders.shape)

display(fact_purchase_orders.head())
```

```python
# MODEL VALIDATION

print("DIMENSIONS")
print("Dim Product:", dim_product["Product_ID"].nunique())
print("Dim Store:", dim_store["Store_ID"].nunique())
print("Dim Warehouse:", dim_warehouse["Warehouse_ID"].nunique())
print("Dim Supplier:", dim_supplier["Supplier_ID"].nunique())
print("Dim Date:", dim_date["Date"].nunique())

print("\nFACT TABLES")
print("Fact Sales:", len(fact_sales))
print("Fact Inventory:", len(fact_inventory))
print("Fact Purchase Orders:", len(fact_purchase_orders))

print("\nPRIMARY KEY CHECKS")

print(
    "Duplicate Sale_ID:",
    fact_sales["Sale_ID"].duplicated().sum()
)

print(
    "Duplicate Order_ID:",
    fact_purchase_orders["Order_ID"].duplicated().sum()
)

print(
    "Duplicate Supplier_ID:",
    dim_supplier["Supplier_ID"].duplicated().sum()
)

print(
    "Duplicate Date:",
    dim_date["Date"].duplicated().sum()
)
```

```python
# SAVE DATA MODEL TABLES

dim_product.to_csv(
    "dim_product.csv",
    index=False
)

dim_store.to_csv(
    "dim_store.csv",
    index=False
)

dim_warehouse.to_csv(
    "dim_warehouse.csv",
    index=False
)

dim_supplier.to_csv(
    "dim_supplier.csv",
    index=False
)

dim_date.to_csv(
    "dim_date.csv",
    index=False
)

fact_sales.to_csv(
    "fact_sales.csv",
    index=False
)

fact_inventory.to_csv(
    "fact_inventory.csv",
    index=False
)

fact_purchase_orders.to_csv(
    "fact_purchase_orders.csv",
    index=False
)

```

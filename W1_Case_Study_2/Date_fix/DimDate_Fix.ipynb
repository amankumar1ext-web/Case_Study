```python
import pandas as pd

sales = pd.read_csv("./sales_cleaned.csv")
```

```python
sales.head()
```

```python
sales['Timestamp'] = pd.to_datetime(sales['Timestamp'])

```

```python
dim_date = pd.DataFrame()

dim_date['FullDate'] = sales['Timestamp'].dt.date
dim_date['Month'] = sales['Timestamp'].dt.month
dim_date['MonthName'] = sales['Timestamp'].dt.month_name()
dim_date['Year'] = sales['Timestamp'].dt.year
```

```python
# Remove duplicates
dim_date = dim_date.drop_duplicates().reset_index(drop=True)

# Generate DateID
dim_date.insert(0, 'DateID', range(1, len(dim_date) + 1))
```

```python
dim_date.head()
```

```python
dim_date.to_csv("dim_date.csv", index=False)
```

```python
sales.head()
```

```python
sales['FullDate'] = sales['Timestamp'].dt.date

sales = sales.merge(
    dim_date[['DateID', 'FullDate']],
    on='FullDate',
    how='left'
)
```

```python
sales[['Timestamp', 'DateID']].head()
```

```python
sales.head()
```

```python
fact_sales = sales[
    [
        'SaleID',
        'CustomerID',
        'ProductID',
        'DateID',
        'Quantity',
        'SalesAmount'
    ]
]
```

```python
fact_sales.to_csv(
    "fact_sales23.csv",
    index=False
)
```

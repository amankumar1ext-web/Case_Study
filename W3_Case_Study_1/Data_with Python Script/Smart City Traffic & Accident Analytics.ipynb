```python
import pandas as pd

# Load datasets
sensor_data = pd.read_csv("./road_traffic_sensor_data.csv")
accident_data = pd.read_csv("./traffic_accident_data.csv")
```

```python
sensor_data.head()
```

```python
accident_data.head()
```

```python
print("SENSOR DATA SHAPE:")
print(sensor_data.shape)

print("\nACCIDENT DATA SHAPE:")
print(accident_data.shape)
```

```python
# Column Names

print("\nSENSOR DATA COLUMNS:")
print(sensor_data.columns.tolist())

print("\nACCIDENT DATA COLUMNS:")
print(accident_data.columns.tolist())
```

```python
print("\nSENSOR DATA INFO:")
sensor_data.info()

print("\nACCIDENT DATA INFO:")
accident_data.info()
```

```python
print("\nSENSOR DATA MISSING VALUES:")
print(sensor_data.isnull().sum())

print("\nACCIDENT DATA MISSING VALUES:")
print(accident_data.isnull().sum())
```

```python
print("\nSENSOR DATA DUPLICATES:")
print(sensor_data.duplicated().sum())

print("\nACCIDENT DATA DUPLICATES:")
print(accident_data.duplicated().sum())
```

```python
# Unique ID validation
print("Sensor IDs:")
print("Total rows:", len(sensor_data))
print("Unique Sensor IDs:", sensor_data["Sensor_ID"].nunique())
```

```python
print("\nAccident IDs:")
print("Total rows:", len(accident_data))
print("Unique Accident IDs:", accident_data["Accident_ID"].nunique())
```

```python
# Date/Time range
print("\nSensor Date-Time:")
print("First:", sensor_data["Date_Time"].min())
print("Last :", sensor_data["Date_Time"].max())

print("\nAccident Date-Time:")
print("First:", accident_data["Date_Time"].min())
print("Last :", accident_data["Date_Time"].max())
```

```python
# Location comparison
print("\nNumber of unique sensor locations:", sensor_data["Location"].nunique())
print("Number of unique accident locations:", accident_data["Location"].nunique())
```

```python
print("\nSensor Locations:")
print(sensor_data["Location"].unique())
```

```python
print("\nAccident Locations:")
print(accident_data["Location"].unique())
```

```python
# Convert Date_Time columns to datetime
sensor_data["Date_Time"] = pd.to_datetime(sensor_data["Date_Time"])
accident_data["Date_Time"] = pd.to_datetime(accident_data["Date_Time"])

# Verify the conversion
print("Sensor Date_Time type:", sensor_data["Date_Time"].dtype)
print("Accident Date_Time type:", accident_data["Date_Time"].dtype)
```

```python
# Check final data types
print("SENSOR DATA TYPES")
print(sensor_data.dtypes)

print("\nACCIDENT DATA TYPES")
print(accident_data.dtypes)

# Check for missing values after conversion
print("\nMissing values after conversion")
print("Sensor:", sensor_data.isnull().sum().sum())
print("Accident:", accident_data.isnull().sum().sum())
```

```python
print("SENSOR DATA — DESCRIPTIVE STATISTICS")
display(sensor_data[["Vehicle_Count", "Average_Speed"]].describe())

print("\nACCIDENT DATA — DESCRIPTIVE STATISTICS")
display(accident_data[["Number_of_Vehicles", "Casualties"]].describe())
```

```python
print("\nSENSOR DATA — CONGESTION LEVEL")
print(sensor_data["Congestion_Level"].value_counts())

print("\nACCIDENT DATA — ACCIDENT SEVERITY")
print(accident_data["Accident_Severity"].value_counts())

print("\nACCIDENT DATA — WEATHER CONDITION")
print(accident_data["Weather_Condition"].value_counts())

print("\nACCIDENT DATA — ROAD CONDITION")
print(accident_data["Road_Condition"].value_counts())

print("\nACCIDENT DATA — TRAFFIC DENSITY")
print(accident_data["Traffic_Density"].value_counts())
```

```python
print("TRAFFIC BY LOCATION")
display(
    sensor_data.groupby("Location")
    .agg(
        Total_Vehicles=("Vehicle_Count", "sum"),
        Average_Speed=("Average_Speed", "mean")
    )
    .sort_values("Total_Vehicles", ascending=False)
)

print("\nACCIDENTS BY LOCATION")
display(
    accident_data.groupby("Location")
    .agg(
        Accident_Count=("Accident_ID", "count"),
        Total_Casualties=("Casualties", "sum")
    )
    .sort_values("Accident_Count", ascending=False)
)
```

```python
# Extract time components
sensor_data["Hour"] = sensor_data["Date_Time"].dt.hour
accident_data["Hour"] = accident_data["Date_Time"].dt.hour

print("TRAFFIC BY HOUR")
display(
    sensor_data.groupby("Hour")
    .agg(
        Total_Vehicles=("Vehicle_Count", "sum"),
        Average_Speed=("Average_Speed", "mean")
    )
    .sort_values("Total_Vehicles", ascending=False)
)

print("\nACCIDENTS BY HOUR")
display(
    accident_data.groupby("Hour")
    .agg(
        Accident_Count=("Accident_ID", "count"),
        Total_Casualties=("Casualties", "sum")
    )
    .sort_values("Accident_Count", ascending=False)
)
```

```python
print("ACCIDENT SEVERITY BY LOCATION")
display(
    pd.crosstab(
        accident_data["Location"],
        accident_data["Accident_Severity"]
    )
)

print("\nTRAFFIC DENSITY BY ACCIDENT SEVERITY")
display(
    pd.crosstab(
        accident_data["Traffic_Density"],
        accident_data["Accident_Severity"]
    )
)
```

```python
# Traffic Volume by Location
import matplotlib.pyplot as plt

traffic_location = (
    sensor_data.groupby("Location")["Vehicle_Count"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))
traffic_location.plot(kind="bar")
plt.title("Total Traffic Volume by Location")
plt.xlabel("Location")
plt.ylabel("Total Vehicle Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Accidents by Location
accident_location = (
    accident_data["Location"]
    .value_counts()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))
accident_location.plot(kind="bar")
plt.title("Accident Count by Location")
plt.xlabel("Location")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Accidents by Weather Condition
weather_counts = accident_data["Weather_Condition"].value_counts()

plt.figure(figsize=(8,5))
weather_counts.plot(kind="bar")
plt.title("Accidents by Weather Condition")
plt.xlabel("Weather Condition")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Traffic Volume and Average Speed by Location
location_analysis = (
    sensor_data.groupby("Location")
    .agg(
        Total_Vehicles=("Vehicle_Count", "sum"),
        Average_Speed=("Average_Speed", "mean")
    )
)

plt.figure(figsize=(8,5))
plt.scatter(
    location_analysis["Total_Vehicles"],
    location_analysis["Average_Speed"]
)

for location, row in location_analysis.iterrows():
    plt.annotate(
        location,
        (row["Total_Vehicles"], row["Average_Speed"])
    )

plt.title("Traffic Volume vs Average Speed by Location")
plt.xlabel("Total Vehicle Count")
plt.ylabel("Average Speed")
plt.tight_layout()
plt.show()
```
